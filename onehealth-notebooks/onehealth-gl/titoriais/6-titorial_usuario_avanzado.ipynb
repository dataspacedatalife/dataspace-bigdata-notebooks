# Como inicio sesión no sistema?

## 1. Activar a VPN do CESGA

Instalación e configuración do software da VPN: https://bigdata.cesga.es/tutorials/vpn.html#/0/1


## 2. Conectarse usando la interfaz web WebUI
Sempre coa VPN activa, conéctate usando o teu nome de usuario e contrasinal do CESGA en https://bigdata.cesga.es/web-ui/app/#/login

![OneHealth DataSpace](../../static/hadoop-web.png)

## 2. Conectarse usando unha conexión ssh

Sempre coa VPN activa, desde unha consola ou un software similar realiza unha conexión SSH usando o teu nome de usuario e contrasinal do CESGA:

ssh nome_usuario@hadoop3.cesga.es

![OneHealth DataSpace](../../static/hadoop-console.png)

# Exemplo: abrir os nosos propios notebooks de Jupyter usando módulos concretos

O sistema permítenos abrir o noso notebook. Unha vez realizada a conexión SSH, podemos abrilo con:

    start_jupyter

Pero, se precisamos módulos concretos de Anaconda, por exemplo, debemos cargalos antes de abrir o notebook:

    module load anaconda3/2018.12
    start_jupyter

Para coñecer os módulos dispoñibles na plataforma, pode usar:

    module available

Máis en: https://bigdata.cesga.es/user-guide/jupyter.html#

## Manuais do sistema

Por favor, consulte os manuais do sistema, xa que conteñen guías de uso sobre as ferramentas dispoñibles na plataforma, así como información sobre a cota de usuario, a migración de datos e diversos titoriais de interese:

https://bigdata.cesga.es/user-guide/overview.html

Por exemplo, entre a información dispoñible pode atopar:
- Información sobre as **cotas de almacenamento** do usuario en Filesystem Quotas.
- Información sobre **Yarn y HDFS**.
- Se utiliza **R** pode interesarlle o titorial de **Sparklyr**: https://bigdata.cesga.es/user-guide/sparklyr.html
- Como empregar **Spark**.
- Lanzar os nosos propios **Jupyter Notebooks** tal e como se mostra no exemplo anterior.
- **Hive** permite consultar datos en HDFS mediante consultas SQL; tamén pode atopar un pequeno titorial. É a opción recomendada para grandes volumes de datos.
- **Impala** tamén permite realizar consultas SQL interactivas sobre datos en HDFS e recoméndase para realizar consultas pequenas, interactivas e rápidas
- Como empregar **Sqoop** para importar facilmente datos desde bases de datos relacionais a HDFS.

Pode atopar os manuais actualizados do sistema Big Data en: https://bigdata.cesga.es/user-guide/quota.html
