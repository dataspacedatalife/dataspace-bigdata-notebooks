# Uso do sistema HDFS (Hadoop Distributed File System)
O sistema HDFS é un sistema de ficheiros baseado en Java utilizado polo noso sistema Big Data. Ofrece almacenamento de datos escalable e fiable, e foi deseñado para abarcar grandes grupos de servidores de mercadorías. Neste caso de uso explicareos algunhas das funcións básicas de xestión de datos ademáis de cómo mover datos entre o supercomputador Finisterrae III (FT3) e o Datalake.

> **ℹ️ PRIMEIRO**: Para poder usar as funcións completas deste titorial primeiro gárdao no teu Home, para elo fai botón dereito sobre o notebook e escolle "Copy", despóis móvete fora do cartafol de "Titoriais" e pégao no teu Home, ou se tes un cartafol para gardar os titoriais no teu Home pégao dentro.

## O almacenamento na plataforma BigData

A plataforma Hadoop do CESGA ten dous sistemas de ficheiros principais:
- HOME
- HDFS

### HOME na plataforma BigData

O directorio HOME é un sistema de ficheiros NFS compartido entre todos os nodos do clúster e non está pensado para ser usado directamente polas aplicacións Spark. Utilízase para:

- Almacenar as configuracións e os ficheiros comúns.
- Almacenamento temporal para cargar os conxuntos de datos antes de movelos a HDFS.
- Almacenamento temporal para recuperar os resultados de HDFS antes de descargalos no ordenador local.

Teña en conta que non ten copia de seguridade e que a HOME da plataforma BigData e a HOME dos usuarios no supercomputador FT3 non son a mesma nin comparten espazo.

### HDFS

O sistema de ficheiros distribuídos de Hadoop (HDFS) é onde almacenas os datos que vas procesar coas aplicacións Spark. Este almacenamento ofrece:

- Mellor rendemento en ficheiros grandes (>100 MB)
- Os ficheiros en HDFS son de tipo escribir unha vez
- HDFS está optimizado para lecturas secuenciais grandes
- Os datos distribúense en varios nodos
- Os ficheiros están divididos en bloques: o tamaño predeterminado do bloque é de 128 MB
- Os bloques replícanse en varios nodos: por defecto 3 réplicas

## As rutas no HDFS

As rutas no HDFS sempre se refiren á Home do usuario no HDFS, así por exemplo se imos subir o ficheiro file.txt e non indicamos máis que o nome do ficheiro no destino entón ese ficheiro gardarase no cartafol:

    /user/username/

Como exemplo usaremos o comando put: así para mover subir un ficheiro dende o cartafol no que esteamos na nosa HOME no FT3 á nosa HOME no sistema BigData usaremos o comando:

``hdfs dfs -put file.txt file.txt``

## FT3: cargar o módulo HDFS:

Conectámonos ó FT3 da maneira habitual e vemos qué require o módulo HDFS para ser cargado con este comando:

``module spider hadoop/3.0.0-cdh6.1.1``

O sistema indica que temos que cargar un dos seguintes módulos antes de poder acceder ós comandos HDFS:

- cesga/2020
- cesga/2022
- cesga/system

Por exemplo cargamos o módulo cesga/2022 co comando:

``module load cesga/2022``

E agora cargamos o módulo relacionado con HDFS:

``module load hadoop/3.0.0-cdh6.1.1``

ou

``module load hadoop``

E xa podemos executar os comandos HDFS, por exemplo para consultar a versión de HDFS do sistema:

``hdfs version``

## Mover datos entre o FT3 e a plataforma BigData usando comando HDFS

Todos estes comandos executaranse unha vez esteamos conectados por ssh ó FT3:

Subir un ficheiro dende a nosa HOME no FT3 á nosa HOME no HDFS:

``hdfs dfs -put local_file.txt remote_file.txt``

Descargar un ficheiro dende HOME no HDFS á nosa HOME no FT3:

``hdfs dfs -get remote_file.txt local_file.txt``

Listar os ficheiros que temos no HDFS, ambos comandos son equivalentes (<current_user> sustitúese polo teu nome de usuario):

``hdfs dfs -ls``
``hdfs dfs -ls /user/<current_user>``

Crear un cartafol na nosa HOME do HDFS:

``hdfs dfs -mkdir testing``

Borrar un cartafol na nosa HOME do HDFS:

``hdfs dfs -rm -r -f testing``
    
Borrar Un ficheiro na nosa HOME do HDFS:

``hdfs dfs -rm testing/file.txt``
    
Ver o contido dun ficheiro:

``hdfs dfs -cat testing/file.txt``

### Os datasets no sistema BigData

Para consultar os datasets dispoñibles podemos consultalos co comando HDFS seguinte tanto dende o FT3 como dende o propio sistema BigData:

``hdfs dfs -ls /datasets/ccmm/``
