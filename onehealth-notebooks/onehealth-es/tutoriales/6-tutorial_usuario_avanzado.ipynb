La plataforma BigData del CESGA ofrece a los participantes de OHDS un entorno listo para usar basado en tecnologías de Big Data como Apache Spark, facilitando el procesamiento y análisis de grandes volúmenes de datos.

# Documentación oficial del sistema Big Data CESGA

Información del sistema y manuales: https://bigdata.cesga.es/


# ¿Cómo conectarse al sistema?

## 1. Activar la VPN del CESGA

Instalación y configuración del software de VPN: https://bigdata.cesga.es/tutorials/vpn.html#/0/1


## 2. Conectarse usando la interfaz web WebUI
Siempre con la VPN activa, conéctate usando tu nombre de usuario y contraseña del CESGA en https://bigdata.cesga.es/web-ui/app/#/login

![OneHealth DataSpace](../../static/hadoop-web.png)

## 2. Conectarse usando una conexión ssh

Siempre con la VPN activa, desde una consola o un software similar realiza una conexión ssh usando tu nombre de usuario y contraseña del CESGA:

ssh nombre_usuario@hadoop3.cesga.es

![OneHealth DataSpace](../../static/hadoop-console.png)

# Ejemplo: abrir nuestros propios notebooks de Jupyter usando módulos concretos

El sistema nos permite abrir nuestro notebook, una vez realizada la conexión ssh podemos abrirlo con:

    start_jupyter

Pero si requerimos módulos concretos de anaconda por ejemplo debemos cargarlos antes de abrir el notebook:

    module load anaconda3/2018.12
    start_jupyter

Para saber los módulos disponibles en la plataforma puede usar:

    module available

Más en: https://bigdata.cesga.es/user-guide/jupyter.html#

## Manuales del sistema

Por favor consulte los manuales del sistema ya que contienen guías de uso sobre las herramientas disponibles en el sistema así como información sobre la cuota del usuario, migración de datos y varios tutoriales interesantes:

https://bigdata.cesga.es/user-guide/overview.html

Por ejemplo, entre la información disponible puede encontrar:
- Información sobre las **cuotas de almacenamiento** del usuoario en Filesystem Quotas.
- Informacińo sobre **Yarn y HDFS**.
- Si utiliza **R** puede interesarle el tutorial de **Sparklyr**: https://bigdata.cesga.es/user-guide/sparklyr.html
- Cómo usar **Spark**.
- Lanzar nuestros propios **Jupyter Notebooks** tal y como se muestra en el ejemplo anterior.
- **Hive** permite consultar datos en HDFS mediante consultas SQL, puede encontrar también un pequeño tutorial. Es la opcioń recomendada para grandes lotes.
- **Impala** también permite realizar consultas SQL interactivas sobre datos HDFS y se recomienda para realizar pequeñas consultas interactivas rápidas.
- Como usar **Sqoop** para importar fácilmente datos desde bases de datos relacionales a HDFS.

Puede encontrar los manuales actualizados del sistema Big Data en: https://bigdata.cesga.es/user-guide/quota.html
