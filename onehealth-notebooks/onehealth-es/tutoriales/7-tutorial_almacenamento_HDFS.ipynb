# Uso del sistema HDFS (Hadoop Distributed File System)
El sistema HDFS es un sistema de archivos basado en Java utilizado por nuestro sistema Big Data. Ofrece un almacenamiento de datos escalable y fiable, y ha sido diseñado para abarcar grandes clústeres de servidores convencionales. En este caso de uso explicaremos algunas de las funciones básicas de gestión de datos, además de cómo mover datos entre el supercomputador Finisterrae III (FT3) y el Data Lake.

## El almacenamiento en la plataforma BigData

La plataforma Hadoop del CESGA tiene dos sistemas de archivos principales:
- HOME
- HDFS

### HOME en la plataforma BigData

El directorio HOME es un sistema de archivos NFS compartido entre todos los nodos del clúster y no está pensado para ser utilizado directamente por las aplicaciones Spark. Se utiliza para:

- Almacenar las configuraciones y los archivos comunes.
- Almacenamiento temporal para cargar los conjuntos de datos antes de moverlos a HDFS.
- Almacenamiento temporal para recuperar los resultados de HDFS antes de descargarlos al ordenador local.

Tenga en cuenta que no dispone de copia de seguridad y que el directorio HOME de la plataforma BigData y el directorio HOME de los usuarios en el supercomputador FT3 no son el mismo ni comparten espacio de almacenamiento.

### HDFS

El Sistema de Archivos Distribuido de Hadoop (HDFS) es donde almacenas los datos que vas a procesar con las aplicaciones Spark. Este almacenamiento ofrece:

- Mejor rendimiento con archivos grandes (>100 MB).
- Los archivos en HDFS son de tipo “escribir una vez”
- HDFS está optimizado para grandes lecturas secuenciales.
- Los datos se distribuyen entre varios nodos.
- Los archivos están divididos en bloques: el tamaño predeterminado de cada bloque es de 128 MB.
- Los bloques se replican en varios nodos: por defecto, se crean 3 réplicas.

## Las rutas en HDFS

Las rutas en HDFS siempre se refieren al directorio Home del usuario en HDFS. Así, por ejemplo, si vamos a subir el archivo file.txt y no indicamos nada más que el nombre del archivo en el destino, entonces ese archivo se guardará en la carpeta:

    /user/username/

Como ejemplo utilizaremos el comando put. Así, para subir un archivo desde el directorio en el que nos encontremos dentro de nuestra carpeta HOME en FT3 a nuestra carpeta HOME en el sistema BigData, utilizaremos el siguiente comando:

``hdfs dfs -put file.txt file.txt``

## FT3: cargar el módulo HDFS:

Nos conectamos a FT3 de la forma habitual y comprobamos qué requisitos tiene el módulo HDFS para poder cargarse mediante el siguiente comando:

``module spider hadoop/3.0.0-cdh6.1.1``

El sistema indica que debemos cargar uno de los siguientes módulos antes de poder acceder a los comandos de HDFS:

- cesga/2020
- cesga/2022
- cesga/system

Por ejemplo, cargamos el módulo cesga/2022 con el siguiente comando:

``module load cesga/2022``

Y ahora cargamos el módulo relacionado con HDFS:

``module load hadoop/3.0.0-cdh6.1.1``

o

``module load hadoop``

Y ya podemos ejecutar los comandos de HDFS. Por ejemplo, para consultar la versión de HDFS del sistema:

``hdfs version``

## Mover datos entre FT3 y la plataforma BigData utilizando comandos HDFS

Todos estos comandos se ejecutarán una vez que estemos conectados por SSH a FT3:

Subir un archivo desde nuestra carpeta HOME en FT3 a nuestra carpeta HOME en HDFS:

``hdfs dfs -put local_file.txt remote_file.txt``

Descargar un archivo desde la carpeta HOME en HDFS a nuestra carpeta HOME en FT3:

``hdfs dfs -get remote_file.txt local_file.txt``

Listar los archivos que tenemos en HDFS. Ambos comandos son equivalentes (<current_user> se sustituye por tu nombre de usuario):

``hdfs dfs -ls``
``hdfs dfs -ls /user/<current_user>``

Crear una carpeta en nuestra HOME de HDFS:

``hdfs dfs -mkdir testing``

Eliminar una carpeta de nuestra HOME en HDFS:

``hdfs dfs -rm -r -f testing``
    
Eliminar un archivo de nuestra HOME en HDFS:

``hdfs dfs -rm testing/file.txt``
    
Ver el contenido de un archivo:

``hdfs dfs -cat testing/file.txt``

### Los datasets en el sistema BigData

Para consultar los datasets disponibles, podemos hacerlo mediante el siguiente comando HDFS, tanto desde FT3 como desde el propio sistema BigData:

``hdfs dfs -ls /datasets/onehealth/``
