# Exportación a csv e descarga dos datos avanzada
Neste **caso de uso** imos explorar distintas opcións de exportación dos datos a ficheiros `csv` e como descargalos no noso equipo.

```python
import onehealth
import pyspark.sql.functions as F
from pyspark.sql.functions import col, expr, min, max, to_date
import pandas as pd
```

```python
afloramiento = onehealth.load("5wqKs2ie8AcFCgX")
```

Pero non precisamos almacenar todas as filas do dataset, podemos por exemplo consultar as primeiras 10 filas utilizando a función `limit`, despóis aplicamos a función `toPandas` para transformar os datos. O notebook mostrará a táboa correspondente ó non indicar máis accións.

Para almacenar estes dátos nun csv no noso directorio, crearemos unha variable onde almacenamos o dataset filtrado. Despóis usamos a función `to_csv`:
- Se non indicamos nada, gardará o ficheiro no mesmo cartafol onde está almaceado o notebook.

```python
# Almacenamos o dataset nun DataFrame de Pandas
descargar = afloramiento.limit(10).toPandas()

# Convertimos e gardamos o novo DataFrame
descargar.to_csv('~/notebook/descargar_csv/afloramiento.csv')
```

Entrando no cartafol <b>"descargar_csv"</b> podemos abrir directamente o ficheiro e o mostrará nunha lapela nova. A exportación por defecto usa `,` como separador.

### Exportar sen cabeceira e sen índices

Por defecto a función `to_csv` exporta os datos incluíndo as cabeceiras, é dicir, os nomes das columnas. Para exportar o csv sen as cabeceiras das columnas usamos o parámetro <b>"header=False"</b>.

```python
descargar.to_csv('~/notebook/descargar_csv/afloramiento_sen_cabecera.csv', header=False)
```

Tamén por defecto inclúe un nome de fila, neste caso un número que identifica a cada fila. Co parámetro <b>"index=False"</b> podemos prescindir desa columna por completo no noso csv. Por exemplo, un ficheiro sen cabeceiras nin índice se xeraría usando os seguintes parámetros:

```python
descargar.to_csv('~/notebook/descargar_csv/afloramiento_sen_cabecera_sen_indice.csv', header=False, index=False)
```

### Exportar a csv comprimindo o ficheiro

A función `to_csv` acepta outros parámetros como son a compresión de ficheiros usando os formatos `zip`, `gzip`, `bz2`, `zstd`, `xz` e `tar`, para elo temos que indicalo usando un diccionario.

Como exemplo:
- Creamos unha variable chamada <b>opc_compresion</b> que sera o noso diccionario cas opciońs que precisa a compresión. Neste exemplo o diccionario indica o método de compresión <b>(method='zip')</b> e o nome do ficheiro csv que se vai comprimir <b>(archive_name='afloramento.csv')</b>. Este ficheiro se xera automáticamente pola función `to_csv`.
- Usamos o diccionario como parámetro da función `to_csv`.

```python
# Creamos o diccionario
opc_compresion = dict(method='zip',
                     archive_name='afloramento_filtro_filas.csv')

# Usamos o diccionario
descargar.to_csv('~/notebook/descargar_csv/afloramiento.zip', compression=opc_compresion)
```

### Exportar a csv usando un separador distinto

O parámetro <b>'sep'</b> permite escoller un separador distinto xa que por defecto é `,`.

```python
descargar.to_csv('~/notebook/descargar_csv/afloramento_separador_punto_e_coma.csv',sep=";")
```

### Usando varias opcións simultáneamente

Podemos, por exemplo, xerar un ficheiro comprimido que conteña un csv que use `;` como separador e non teña cabeceiras.

```python
# Creamos o diccionario
opc_compresion = dict(method='zip',
                     archive_name='afloramento_separador_sencabeceira.csv')

# Usamos o diccionario
descargar.to_csv('~/notebook/descargar_csv/afloramento_separador_sencabeceira.zip', header=False, sep=";", compression=opc_compresion)
```

### Crear un csv despóis de filtrar os datos

Podemos usar as opcións de Pandas para filtrar os nosos datos e almacenar o resultado de ditos filtros nun csv. Neste caso temos varias condiciós no noso filtro:
- `water_u` ten que ter valor e este ten que ser maior que 0.0. Como exemplo seleccionamos a columna a filtrar usando o formato `nome_dataframe.nome_columna`.
- A latitude e a lonxitude teñen que coindicir cas indicadas. Como exemplo seleccionamos a columna a filtrar usando a función `col(nome_columna)`.

Filtramos o conxunto e datos buscando aquelas filas que conteñan algo na columna `water_u` e que teñan unha latitude en concreto:

```python
filtrado = afloramiento.filter((afloramiento.water_u!='none') & (afloramiento.water_u>0.0) & (col('latitude')==40.78689956665039) & (col('longitude')==-9.502409934997559))
```

Podemos seguir aplicando filtros cando se precise. Por exemplo, neste caso soamente mostramos as filas con `site_status==1`.

```python
filtrado = filtrado.filter(col('site_status')==1)
```

Deses datos xa filtrados vamos a escoller un conxunto de columnas:

```python
columnas = filtrado [["time","latitude","longitude","water_u"]]

# Pasamos as primeiras 100 filas a Pandas
columnas = columnas.limit(100).toPandas()
```

Almacenamos este resultado nun csv usando o xa aprendido:

```python
columnas.to_csv('~/notebook/descargar_csv/afloramento_filtrado.csv', index=False, sep=";")
```

Entrando no cartafol <b>"descargar_csv"</b> podemos abrir directamente o ficheiro e o mostrará nunha lapela nova.
