# Exportación a csv e descarga dos datos
Neste **caso de uso** imos explorar distintas opcións de exportación dos datos a ficheiros `csv` e como descargalos no noso equipo. 

```python
import onehealth
import pyspark.sql.functions as F
from pyspark.sql.functions import col, expr, min, max, to_date
import pandas as pd
```

```python
capturas = onehealth.load("SS4jXDXK2DgTDLE")
```

Para almacenar estes dátos nun csv no noso directorio, crearemos unha variable onde almacenamos o dataset filtrado. Despóis usamos a función `to_csv`:
- Se non indicamos nada, gardará o ficheiro no mesmo cartafol onde está almaceado o notebook.

```python
# Almacenamos o dataset nun DataFrame de Pandas
descargar = capturas.limit(10).toPandas()

# Creamos o directorio para almacenar os ficheiros
!mkdir ~/notebook/descargar_csv

# Convertimos e gardamos o novo DataFrame
descargar.to_csv('~/notebook/descargar_csv/exemplo_capturas.csv')
```

Entrando no cartafol <b>"descargar_csv"</b> podemos abrir directamente o ficheiro e o mostrará nunha lapela nova. A exportación por defecto usa "," como separador.

### Descarga dos ficheiros no noso equipo

Para descargar os ficheiros xerados ca exportacioń a csv (sexan csv ou comprimidos), soamente temos que facer click co botón dereito do rato sobre o ficheiro no panel de navegación situado á esquerda e seleccionar "Download". Podemos descargar varios ficheiros á vez se os seleccionamos todos.
